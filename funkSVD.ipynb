```python
import os
from surprise import Dataset, BaselineOnly, Reader, SVD, accuracy
from surprise.model_selection import cross_validate, train_test_split
```

```python
file_path = os.path.abspath("./ml-100k/u.data")
reader = Reader(line_format='user item rating timestamp', sep='\t')
data = Dataset.load_from_file(file_path, reader=reader)
data.raw_ratings[:5]
```

```python
trainset, testset = train_test_split(data, test_size=0.25)

#Parametros por defecto
# n_factors = 100
# n_epochs = 20
# lr_all = 0.005 (learning rate para todos los parámetros)
# reg_all = 0.02 (Término de regularización para todos los parámetros)

algo = SVD()
algo.fit(trainset)
```

```python
predictions = algo.test(testset)
accuracy.rmse(predictions)
```
